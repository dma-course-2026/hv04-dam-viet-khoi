```python
# Bai 9 - Tinh tien dien
def tinhtiendien():
    hoten = input("Ho va ten: ")
    chiso1, chiso2 = map(int, input("Nhap chi so thang truoc, chi so thang nay: ").split())
    sodien = chiso2 - chiso1
    # Tinh tien theo bac
    if sodien <= 60:
        tien = sodien * 5
    elif sodien <= 160:
        tien = 60 * 5 + (sodien - 60) * 8
    else:
        tien = 60 * 5 + 100 * 8 + (sodien - 160) * 10
    print("Ho va ten:", hoten)
    print("Tien phai tra la %8.2f" % tien)
    return tien
tinhtiendien()
```

```python
# Bai 14 - Hinh tam giac
import math
def tamgiac():
    a, b, c = map(float, input("Nhap 3 so a,b,c = ").split())
    # Kiem tra 3 canh tam giac
    if a + b > c and a + c > b and b + c > a:
        chuvi = a + b + c
        p = chuvi / 2
        # cong thuc Heron
        dientich = math.sqrt(p * (p - a) * (p - b) * (p - c))
        print("Chu vi tam giac:", chuvi)
        print("Dien tich tam giac:", dientich)
    else:
        print(a, ",", b, ",", c, "Khong phai la 3 canh cua tam giac")
tamgiac()
```

```python
# Bai 20 - Phuong trinh bac 2
import math
def ptbac2():
    a = float(input("nhap a: "))
    b = float(input("nhap b: "))
    c = float(input("nhap c: "))
    delta = b**2 - 4*a*c
    if delta < 0:
        print("Phuong trinh vo nghiem")
    elif delta == 0:
        print("Phuong trinh co nghiem kep x =", -b / (2*a))
    else:
        x1 = (-b + math.sqrt(delta)) / (2*a)
        x2 = (-b - math.sqrt(delta)) / (2*a)
        print("Phuong trinh co 2 nghiem phan biet:")
        print("x1 =", x1)
        print("x2 =", x2)
    return delta
ptbac2()
```

```python
# Bai 32 - He phuong trinh bac 1 hai an
def hept():
    a = float(input("Nhap a = "))
    b = float(input("Nhap b = "))
    m = float(input("Nhap m = "))
    c = float(input("Nhap c = "))
    d = float(input("Nhap d = "))
    n = float(input("Nhap n = "))
    # Cramer
    D = a*d - b*c
    Dx = m*d - b*n
    Dy = a*n - c*m
    if D != 0:
        print("He phuong trinh co 1 cap nghiem:")
        print("x =", Dx / D, " va  y =", Dy / D)
    elif Dx == 0 and Dy == 0:
        print("He phuong trinh vo so nghiem")
    else:
        print("He phuong trinh vo nghiem")
    return D
hept()
```

```python
# Bai 33 - Diem thuoc duong tron
import math
def diemduongtron():
    R = float(input("Nhap ban kinh R: "))
    print("Nhap toa do tam duong tron:")
    x0 = float(input("x0: "))
    y0 = float(input("y0: "))
    print("Nhap toa do diem A:")
    xa = float(input("xa: "))
    ya = float(input("ya: "))
    # khoang cach tu A toi tam
    d = math.sqrt((xa - x0)**2 + (ya - y0)**2)
    if d < R:
        print("Diem A nam trong duong tron")
    elif d == R:
        print("Diem A nam tren duong tron")
    else:
        print("Diem A nam ngoai duong tron")
    return d
diemduongtron()
```

```python
# Bai 45 - Kiểm tra số nguyên tố
def ktnguyento():
    n = int(input("Nhap so n: "))
    nguyento = True
    if n < 2:
        nguyento = False
    # chỉ cần xét tới căn n
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            nguyento = False
            break
    if nguyento:
        print(n, "la so nguyen to")
    else:
        print(n, "khong phai so nguyen to")
    return nguyento
ktnguyento()
```
