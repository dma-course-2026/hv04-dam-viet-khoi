```python
# Bai 1 - Chia het cho 7 khong chia het cho 5
def tim7():
    ketqua = []
    for i in range(2000, 3201):
        if i % 7 == 0 and i % 5 != 0:
            ketqua.append(str(i))
    print(",".join(ketqua))
    return ketqua
tim7()
```

```python
# Bai 2 - Giai thua
class GiaiThua:
    def tinh(self, n):
        kq = 1
        for i in range(2, n + 1):
            kq = kq * i
        return kq
n = int(input("Nhap so can tinh giai thua: "))
gt = GiaiThua()
print(gt.tinh(n))
```

```python
# Bai 9 - Tinh tuoi tu ngay sinh
import datetime
def tinhtuoi():
    ngaysinh = input("Nhap ngay sinh (y/m/d): ")
    ds = ngaysinh.split("/")
    nam = int(ds[0])
    thang = int(ds[1])
    ngay = int(ds[2])
    homnay = datetime.date.today()
    tuoi = homnay.year - nam
    # chua toi sinh nhat thi tru 1
    if (homnay.month, homnay.day) < (thang, ngay):
        tuoi = tuoi - 1
    print("Tuoi la:", tuoi)
    return tuoi
tinhtuoi()
```

```python
# Bai 10 - Tinh luong thuc linh
def tinhluong():
    sogio = float(input("Nhap so gio lam moi tuan: "))
    luonggio = float(input("Nhap thu lao moi gio: "))
    if sogio <= 44:
        luong = sogio * luonggio
    else:
        luong = 44 * luonggio + (sogio - 44) * luonggio * 1.5
    print("Thuc linh:", luong)
    return luong
tinhluong()
```

```python
# Bai 11 - Q = sqrt(2*C*D/H)
import math
def tinhq():
    c = 50
    h = 30
    dayd = input("Nhap cac gia tri D, cach nhau bang dau phay: ").split(",")
    ketqua = []
    for d in dayd:
        q = math.sqrt(2 * c * int(d) / h)
        ketqua.append(str(round(q)))
    print(",".join(ketqua))
    return ketqua
tinhq()
```

```python
# Bai 18 - Dem chu cai va chu so
def demchuso():
    cau = input("Nhap cau: ")
    sochucai = 0
    sochuso = 0
    for kitu in cau:
        if kitu.isalpha():
            sochucai += 1
        elif kitu.isdigit():
            sochuso += 1
    print("So chu cai la:", sochucai)
    print("So chu so la:", sochuso)
    return sochucai, sochuso
demchuso()
```

```python
# Bai 22 - So du tai khoan ngan hang
def taikhoan():
    sodu = 0
    print("Nhap nhat ky giao dich, nhap dong trong de dung:")
    while True:
        dong = input()
        if dong == "":
            break
        loai, tien = dong.split()
        if loai == "D":
            sodu += int(tien)
        elif loai == "W":
            sodu -= int(tien)
    print(sodu)
    return sodu
taikhoan()
```

```python
# Bai 23 - Kiem tra va bien doi day doi xung
def kiem_tra_doi_xung(day):
    return day == day[::-1]

def sx_doi_xung(day):
    n = len(day)
    for i in range(n // 2):
        if day[i] != day[n - 1 - i]:
            day[n - 1 - i] = day[i]
    return day

def doixung():
    day = []
    for x in input("Nhap day so nguyen, cach nhau bang khoang trang: ").split():
        day.append(int(x))
    if kiem_tra_doi_xung(day):
        print("Day da doi xung:", day)
    else:
        print("Day chua doi xung:", day)
        sx_doi_xung(day)
        print("Day sau khi bien doi:", day)
doixung()
```
